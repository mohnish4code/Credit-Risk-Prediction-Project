```python
features = ['loan_amnt', 'loan_int_rate', 'loan_percent_income']

Q1 = df[features].quantile(0.25)
Q3 = df[features].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df[features] < (Q1 - 1.5 * IQR)) | 
          (df[features] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

**Now we have obtained which we are going to use for the training affecting the target feature mainly**

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop="first", sparse_output=False)
```

```python
encoded = ohe.fit_transform(df[["loan_intent", "person_home_ownership"]])
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(["loan_intent", "person_home_ownership"])
)
```

```python
df = df.drop(["loan_intent", "person_home_ownership"], axis="columns")
```

```python
df = pd.concat([df, encoded_df], axis="columns")
```

```python
df.head()
```

**Now the encoding part is finished now we have to drop the unnecessary features**

```python
df = df.drop(["person_age", "person_income", "person_emp_length", "cb_person_default_on_file",
              "cb_person_cred_hist_length","loan_grade"], axis="columns")
```

```python
df.head()
```

```python
X = df.drop("loan_status", axis="columns")
y = df.loan_status
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=44)
```
